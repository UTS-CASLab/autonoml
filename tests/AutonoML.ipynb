Finding the metrics values

```python
from sklearn.metrics import r2_score, mean_absolute_percentage_error

# Load the CSV file
df = pd.read_csv("results/responses.csv")

# Extract the true and predicted values
y_true = df['T_Y']
y_pred = df['S_Y']

# Calculate R-squared
r2 = r2_score(y_true, y_pred)

# Calculate MAPE
mape = mean_absolute_percentage_error(y_true, y_pred)

# Print results
print(f"R²: {r2:.4f}")
print(f"MAPE: {mape:.4%}")
```
